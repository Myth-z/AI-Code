```python
from typing import List

def split_into_chunks(doc_file: str) -> List[str]:
    with open(doc_file, 'r', encoding = 'utf-8') as file:
        content = file.read()
    return [chunk for chunk in content.split("\n\n")]
```

```python
chunks = split_into_chunks("doc.md")

##切片
for i, chunk in enumerate(chunks):
    print(f"[{i}] {chunk}\n")
```

```python
from sentence_transformers import SentenceTransformer

embedding_model = SentenceTransformer("shibing624/text2vec-base-chinese")

##把切片向量化
def embed_chunk(chunk: str) -> List[float]:
    embedding = embedding_model.encode(chunk)
    return embedding.tolist()

test_embedding = embed_chunk("测试内容")
print(len(test_embedding))
print(test_embedding)
```

```python
##把所有切片转化为向量
embeddings = [embed_chunk(chunk) for chunk in chunks]
print(len(embeddings))
print(embeddings[0])
```

```python
import chromadb

chromadb_client = chromadb.PersistentClient("./chroma.db")
chromadb_collection = chromadb_client.get_or_create_collection(name="default")

##把向量化后的切片保存到向量数据库中
def save_embeddings(chunks: List[str], embeddings:List[List[float]]) -> None:
    ids = [str(i) for i in range(len(chunks))]
    chromadb_collection.add(
        documents = chunks,
        embeddings = embeddings,
        ids = ids
    )

save_embeddings(chunks, embeddings)
```

```python
##根据用户提问，进行数据召回
def retrieve(query: str, top_k: int) -> List[str]:
    query_embedding = embed_chunk(query) ##先把问题也转化为向量
    results = chromadb_collection.query( ##使用chromadb查询top_k个最相似的结果
        query_embeddings = [query_embedding],
        n_results = top_k
    )

    return results['documents'][0]

query = "哆啦A梦使用的3个秘密道具是什么？" ##进行测试
retrieved_chunks = retrieve(query, 5)

for i, chunk in enumerate(retrieved_chunks):
    print(f"[{i}]{chunk}\n")
```

```python
from sentence_transformers import CrossEncoder

##重排序
def rerank(query: str, retrieved_chunks: List[str], top_k: int) -> List[str]:
    cross_encoder = CrossEncoder('cross-encoder/mmarco-mMiniLMv2-L12-H384-v1')
    pairs = [(query, chunk) for chunk in retrieved_chunks]
    scores = cross_encoder.predict(pairs)

    ##查看每个分块对应的分数，并排序
    chunk_with_score_list = [(chunk, score)
                            for chunk, score in zip(retrieved_chunks, scores)]
    chunk_with_score_list.sort(key = lambda pair: pair[1], reverse = True)

    return [chunk for chunk, _ in chunk_with_score_list][:top_k]

reranked_chunks = rerank(query, retrieved_chunks, 3)

for i, chunk in enumerate(reranked_chunks):
    print(f"[{i}] {chunk}\n")
```

```python
from dotenv import load_dotenv
import dashscope
from dashscope import Generation
import os

##利用大模型结合切片内容回答用户提问
load_dotenv()
dashscope.api_key = os.getenv('DASHSCOPE_API_KEY')

def generate(query: str, chunks: List[str]) -> str:
    separator = "\n\n" 
    prompt = f"""你是一个知识助手，请根据用户提问和下列片段生成准确的回答，口吻要正经严肃，并给出信息源。

用户问题：{query}

相关片段：
{separator.join(chunks)}

请基于上列信息回答问题，不要编造虚假信息。"""

    print(f"{prompt}\n\n---\n")

    try:
        response = Generation.call(
            model="qwen-plus",
            messages=[{'role': 'user', 'content': prompt}]
        )

        if response.status_code != 200:
            return f"API Error: {getattr(response, 'message', 'Unknown error')} (Code: {response.status_code})"

        return response.output.text

    except Exception as e:
        return f"Exception: {str(e)}"

answer = generate(query, reranked_chunks)
print(answer)
```
